# One CellProfiler export

{func}`~mantispy.ds.jump_export` is a single, unmodified `ExportToSpreadsheet` directory, one field of view of a DMSO well of `BR00121438` {cite:p}`Chandrasekaran_2023`. It is the example input for {func}`~mantispy.io.read_profiles`: the raw `Image.csv` plus the `Cells`, `Cytoplasm` and `Nuclei` tables, exactly as CellProfiler wrote them. The images this was measured from are in {func}`~mantispy.ds.jump_plate`, and the well-level profiles of the same plate in {func}`~mantispy.ds.jump_target2`.

**Use it if** you have your own CellProfiler `ExportToSpreadsheet` output and want to read it into cells. [From a CellProfiler run](../tutorials/data/cellprofiler.ipynb) walks through that on this directory.

```python
import mantispy as mt
```

```python
directory = mt.ds.jump_export()
sorted(p.name for p in directory.iterdir())
```

## Reading it into cells

`Cytoplasm` is the only object the JUMP pipeline gives a parent in both `Cells` and `Nuclei`, so it is the one that joins all three tables, and one cytoplasm is one cell.

```python
import warnings

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    cells = mt.io.read_profiles(directory, primary_object="Cytoplasm", resolution="cell")
cells
```

The metadata columns that came off the export name each cell by its plate, well, site and object number.
